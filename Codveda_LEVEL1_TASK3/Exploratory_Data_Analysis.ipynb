```python
import pandas as pd
import numpy as np

# Load our preprocessed dataset from Task 2
df_clean = pd.read_csv("codveda_level1_task2_preprocessed.csv")

print("--- Deep Summary Statistics ---")
# Compute mean, median (50%), std, min, max, and IQR
summary_stats = df_clean.describe()
print(summary_stats)

print("\n--- Feature Skewness ---")
# Skewness measures the asymmetry of the probability distribution
numeric_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
print(df_clean[numeric_cols].skew())

print("\n--- Correlation Matrix ---")
# Correlation matrix to check linear relationships between features
corr_matrix = df_clean[numeric_cols].corr()
print(corr_matrix)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set professional plotting style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 1. Correlation Heatmap
corr = df_clean[['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species_encoded']].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5, ax=axes[0])
axes[0].set_title("Feature Correlation Heatmap", fontsize=14, fontweight='bold')

# 2. Pairwise Distribution Boxplot
sns.boxplot(data=df_clean[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']], ax=axes[1], palette="Set2")
axes[1].set_title("Feature Distributions (Post-Outlier Removal & Scaling)", fontsize=14, fontweight='bold')
axes[1].set_ylabel("Standardized Value")

plt.tight_layout()
plt.show()

print("EDA Visualizations generated successfully! Level 1 is fully complete.")
```
