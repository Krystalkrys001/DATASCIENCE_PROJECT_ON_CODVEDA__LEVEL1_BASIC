```python
import pandas as pd
import numpy as np

# Load the Iris dataset
df_iris = pd.read_csv('1) iris.csv')

print("--- Raw Dataset Info ---")
print(df_iris.info())

print("\n--- First 5 Rows ---")
print(df_iris.head())

print("\n--- Summary Statistics ---")
print(df_iris.describe())
```

```python
from sklearn.preprocessing import StandardScaler

# 1. Outlier Detection & Removal using IQR for 'sepal_width' (which has visible spread)
Q1 = df_iris['sepal_width'].quantile(0.25)
Q3 = df_iris['sepal_width'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter dataset to remove outliers for sepal_width
df_filtered = df_iris[(df_iris['sepal_width'] >= lower_bound) & (df_iris['sepal_width'] <= upper_bound)]
print(f"Original rows: {len(df_iris)} | Rows after outlier filtering: {len(df_filtered)}")

# 2. Feature Standardization (StandardScaler)
# Separate features and target species
X = df_filtered.drop(columns=['species'])
y = df_filtered['species']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for clean preview
df_scaled = pd.DataFrame(X_scaled, columns=X.columns)
df_scaled['species'] = y.values

print("\n--- Standardized Feature Preview (Mean=0, Std=1) ---")
print(df_scaled.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
encoder = LabelEncoder()

# Encode the 'species' column into numerical classes
df_scaled['species_encoded'] = encoder.fit_transform(df_scaled['species'])

print("--- Categorical Encoding Mapping ---")
for class_name, encoded_val in zip(encoder.classes_, range(len(encoder.classes_))):
    print(f"Species '{class_name}' -> Encoded Value {encoded_val}")

print("\n--- Final Preprocessed Dataset Preview ---")
print(df_scaled[['sepal_length', 'sepal_width', 'species', 'species_encoded']].head())

# Export the clean, preprocessed dataset for your portfolio repository
df_scaled.to_csv("codveda_level1_task2_preprocessed.csv", index=False)
print("\nSuccessfully exported preprocessed dataset as 'codveda_level1_task2_preprocessed.csv'!")
```
